# 플랫폼과 모델 B별 특성 t 검정

원본 최종 수정 기록: 2025-06-01T21:04:00+09:00.

파일명의 번호는 파이프라인 순서이며 버전 번호가 아닙니다. 데이터·모델 경로와 실행 환경을 설정한 뒤 필요한 단계를 실행하세요.


```python
import pandas as pd
```

```python
df = pd.read_csv('train_Z_score.csv')
```

```python
from scipy.stats import ttest_ind

# 봇과 휴먼 계정으로 나누기
bot_df = df[df['fake'] == 1]
human_df = df[df['fake'] == 0]

# t-test 수행
t_test_results = []
for col in df.columns[:-1]:  # 'fake' 제외
    t_stat, p_value = ttest_ind(bot_df[col], human_df[col], equal_var=False)
    t_test_results.append({
        "feature": col,
        "t_statistic": t_stat,
        "p_value": p_value,
        "reject_null": p_value < 0.05
    })

# 결과를 DataFrame으로 변환하고 정렬
t_test_df = pd.DataFrame(t_test_results).sort_values(by="p_value")

# CSV로 저장
t_test_df.to_csv("insta_t_test_results.csv", index=False)
```
